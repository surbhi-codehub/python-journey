```python
#1.
def welcome(name):
  return "Welcome, " + name + "!"

result = welcome("Surbhi")
print(result)
```

```python
#2.
def larger(a, b):
  if a > b:
    return a
  else:
    return b

print(larger(10, 20))
```

```python
#3.
def city(name="Ranchi"):
  print("I live in", name)

city()
city("Jamshedpur")
```

```python
#4.
def total(*args):
  return sum(args)


print(total(10, 20, 30))
print(total(5, 10, 15, 20))
```

```python
#5.
def details(**kwargs):
  for key, value in kwargs.items():
    print(key, ":", value)

details(name="Surbhi", age=20, city="Ranchi")
```

```python
#6.
def information(name, age):
  return f"My name is {name} and I am {age} years old."


print(information("Surbhi", 20))
```

```python
#7.
def average(m1, m2, m3,):
  return (m1 + m2 + m3)

print(average(80, 70,90))
```

```python
#8.
def average(m1, m2, m3):
  return (m1 + m2 + m3) / 3

print(average(80, 70, 90))
print(average(60, 75, 85))
print(average(90,95,100))
```

```python
#9.
def add(a, b):
  return a + b

result = add(10, 20)

print(result)
```

```python
#10.
def student(name, age):
  print("Name:", name)
  print("Age:", age)

student(age=20, name="Surbhi")
```
