```python
#3 Print numbers from 1 up to the entered value using a for loop.
num = int(input("Enter a number: "))

for i in range(1, num + 1):
  print(i)
```

```python
#4 Print the same numbers in revers using a while loop
num = int(input("Enter a number: "))

i = num

while i >= 1:
  print(i)
  i -= 1
```

```python
#5 Use the break statement to stop aloop at chosen number.
num = int(input("Enter a number: "))
stop = int(input("Enter the stop number: "))

for i in range(1, num + 1):
  if i == stop:
    break
  print(i)
```

```python
#6 Use the continue statment to skip one chose number.
num = int (input("Enter a number: "))
skip = int(input("Enter the numbr to skip: "))

for  i in range(1, num + 1):
   if i == skip:
    continue
   print(i)
```
