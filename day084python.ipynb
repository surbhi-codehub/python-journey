```python
#1.
def add(a, b):
  return a + b

print(add(10, 20))
```

```python
#2.
def greet(name):
  return "Hello " + name

print(greet("Surbhi"))
```

```python
#3.
def larger(a, b, c):
  return max(a, b, c)

print(larger(10, 25, 15))
```

```python
#4.
def cube(n):
  return n ** 3

print(cube(4))
```

```python
#5.
def uppercase(word):
  return word.upper()

print(uppercase("Python"))
```

```python
#6.
def multiply(a, b):
  return a * b

print(multiply(5, 6))
```
