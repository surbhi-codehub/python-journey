```python
#1. Write a for loop to count down from 10 to 1.
for i in range(10, 0, -1):
  print(i)
```

```python
#2 Print each number on a new line.
for i in range(10, 0, -1):
  print(i)
```

```python
#3 Change the starting number to 5.
for i in range(5, 0, -1):
  print(i)
```

```python
#4 Count down from 20 to 10.
for i in range(20, 9, -1):
   print(i)
```

```python
#5 Try a different step value while counting down.
for i in range(20, 0, -2):
  print(i)
```

```python
#6 Observe how reverse loops work.
for i in range(10, 0, -1):
  print(i)
```

```python
#7 Find the sum of numbers from 1 to 10 using a for loop
total = 0

for i in range(1, 11):
    total = total + i

print(total)
```
