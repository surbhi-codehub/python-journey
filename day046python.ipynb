```python
#1 Write a simple for loop.
for i in range(1, 6):
  print(i)
```

```python
#2 Use the continue statment to skip one number.
for i in range(1, 6):
  if i == 3:
    continue
  print(i)
```

```python
#3 Print the remaining numbers.
for i in range(1, 11):
  if i == 3:
    continue
  print(i)
```

```python
#4 Change the skipped number and observe the output.
for i in range(1, 11):
  if i == 7:
    continue
  print(i)
```

```python
#5 Try the continue statement inside a while loop.
i = 1

while i <= 5:
  if i == 3:
    i += 1
    continue
  print(i)
  i += 1
```
