```python
#1.
def square(num):
  return num * num

print(square(5))
```

```python
#2.
def cube(num):
  return num * num * num

print(cube(5))
```

```python
#3.
def square(num):
  return num * num

def cube(num):
  return num * num * num

num = 5

print("square:", square(num))
print("cube:", cube(num))
```

```python
#4.
def result(marks):
  if marks >= 40:
    return "pass"
  else:
    return "fail"

print(result(75))
print(result(30))
```

```python
#5.
def perimeter(length, width):
  return 2 * (length + width)

print(perimeter(10, 5))
```

```python
#6.
def average(*args):
  return sum(args) / len(args)

print(average(10, 20,30, 40, 50))
```

```python
#7.
def student_details(**kwargs):
   for key, value in kwargs.items():
        print(key, ":", value)

student_details(
    name="Surbhi",
    age=20,
    cource="BCA",
    city="Ranchi"
)
```
