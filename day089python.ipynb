```python
# Topic: Variable-Length Argument(*args)
```

```python
#1.
def names(*args):
  print(args)


names("Rahul", "Aman", "Priya")
```

```python
#2.
def names(*args):
  for name in args:
    print(name)

names("Rahul", "Aman", "Priya")
```

```python
#3.
def names(*args):
  for name in args:
    print(name)


names("Rahul", "Aman", "Priya", "Neha", "Riya")
```

```python
#4.
def numbers(*args):
  for num in args:
    print(num)


numbers(10, 20, 30, 40, 50)
```

```python
#5.
def total(*args):
  return sum(args)


print(total(10, 20, 30, 40))
```

```python
#6.
def count_values(*args):
  return len(args)


print(count_values(10, 20, 30, 40, 50))
```

```python
#7.
def first_value(*args):
  print(args[0])


first_value(10, 20, 30, 40)
```

```python
#8.
def display (*args):
  for value in args:
    print(value)


display("Apple", "Mango", "Banana", "Orange")

```

```python
#9.
def total(*args):
  return sum(args)

result = total(10, 20, 30, 40, 50)

print(result)
```

```python
#10.
def student(name, age, city):
  print("Name:", name)
  print("Age:", age)
  print("City:", city)


student(name="Rahul", age=20, city="Ranchi")
```
