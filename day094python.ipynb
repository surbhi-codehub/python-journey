```python
#1.
def area_of_circle(radius):
  pi = 3.14
  return pi * radius * radius

result = area_of_circle(5)
print(result)
```

```python
#2.
def can_vote(age):
  if age >= 18:
    return "Can vote"
  else:
    return "Cannot vote"

result = can_vote(19)
print(result)
```

```python
#3.
def convert_time(minutes):
  hours = minutes // 60
  remaining_minutes =minutes % 60
  return hours, remaining_minutes

hours, minutes = convert_time(120)

print("Hours:", hours)
print("Minutes:", minutes)
```

```python
#4.
def final_price(price, discount):
  discount_amount = price * discount / 100
  return price - discount_amount

result = final_price(1000, 20)
print("Final Price:", result)
```

```python
#5.
def convert_lowercase(word):
  return word.lower()

result = convert_lowercase("PYTHON")
print(result)
```

```python
#6.
def highest_number(*args):
  return max(args)

result = highest_number(10, 25, 7, 40, 15)
print("Highest Number:", result)
```

```python
#7.
def employee_details(**kwargs):
  for key, value in kwargs.items():
    print(key, ":", value)

employee_details(
  name="Rahul",
  age=25,
  department="IT",
  salarly=30000
)
```

```python
#8.
def add():
  return 10 + 5

def multiply():
  return 10 * 5

def show_result():
  print("Addition:", add())
  print("Multiplication:", multiply())

show_result()
```

```python
#9.
def square(number):
  return number * number

result = square(6)

print("square:", result)
```

```python
#10.
def larger_value(a, b):
  if a > b:
    return a
  else:
    return b

result = larger_value(25, 40)
print("Larger Value:", result)
```
