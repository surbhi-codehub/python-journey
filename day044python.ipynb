```python
#1 write a simple while loop.
i = 1

while i < 5:
    print(i)
    i += 1
```

```python
#2 Observe what happens when the loop condition never changes.
i = 1

while i <= 10:
  print(i)
  i += 1
```

```python
#3 Stop the program safely.
while True:
  print("Hello")
  break
```

```python
#4 Add a count to prevent an infinite loop.
count = 1

while count <= 5:
    print(count)
    count +=1
```

```python
#5 Print numbers from 1 to 5 using the corrected loop.
num = 1

while num <= 5:
    print(num)
    num+= 1
```

```python
#6 Change the ending value and observe the output.
num = 1

while num <= 10:
  print(num)
  num += 1
```

```python
#7 Print the multiplication table of anumber using a while loop.
num = 5
i = 1

while i <= 10:
  print(num, "x", i, "=", num * i)
  i += 1
```
