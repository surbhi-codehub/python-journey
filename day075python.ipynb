```python
#1.
for i in range(5):
  for j in range(5):
    if i == j:
      print("#", end = " ")
    else:
      print("*", end=" ")
  print()
```

```python
#2.
for i in range(1, 6):
    for j in range(1, 6):
        num = i * j
        if num % 2 == 0:
            print(num, end=" ")
    print()
```

```python
#3.
for i in range(1, 7):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
#4.
for row in range(1, 6):
    for column in range(1, 6):
      if row == column:
          print("Row:", row, "Column:", column)
```

```python
#5.
for i in range(5):
    for j in range(5):
        if i == 0 or i == 4 or j == 0 or j == 4:
            print("X", end=" ")
        else:
            print("0", end=" ")
    print()
```

```python
#6.
for i in range(2, 6):
  print("Table of", i)


  for j in range(1, 11):
      print(i, "X", j, "=", i * j)

  print()
```

```python
#7.
for row in range(1, 6):
    for column in range(1, 6):
      if row > column:
          print("(", row, ",", column, ")")
```

```python
#8.
for i in range(1, 6):
  for j in range(1, 6):
    print(j, end=" ")
  print()
```

```python
#9.
for i in range(5):
  for j in range(5):
    if (i + j) % 2 == 0:
      print("x", end=" ")
    else:
      print("0", end=" ")
  print()
```

```python
#10.
for i in range(1, 6):
  for j in range(i):
    print("*", end=" ")
  print()
```
