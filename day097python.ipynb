```python
#1.
def total_marks(s1, s2, s3, s4, s5):
  total = s1 + s2 + s3 + s4 + s5
  return total

result = total_marks(34, 43, 44, 77, 12)

print("Total Marks:", result)
```

```python
#2.
def average(total):
  avg = total / 5
  return avg

result = average(434)

print("Average:", result)
```

```python
#3.
def calculate_grade(avg):
  if avg >= 90:
    return "A+"
  elif avg >= 80:
    return "A"
  elif avg >= 70:
    return "B"
  elif avg >= 60:
    return "C"
  else:
    return "Fail"

result = calculate_grade(85)

print("Grade:", result)
```

```python
#4.
def student_result(name, total, avg, grade):
  print("Student Name:", name)
  print("Total Marks:", total)
  print("Average:", avg)
  print("Grade:", grade)
student_result("Rahul", 454, 53, "A")
```

```python
#5.
def convert_days(days):
  weeks = days // 7
  remaining_days = days % 7
  return weeks, remaining_days

result = convert_days(25)

print("Weeks:", result[0])
print("Remanining Days:", result[1])
```

```python
#6.
def highest_number(*args):
  return max(args)

result = highest_number(10, 50, 32, 78, 56)

print("Highest Number:", result)
```

```python
#7.
def employee_info(**kwargs):
  for key, value in kwargs.items():
    print(key, ":", value)

employee_info(
    name="Rahul",
    age=34,
    salary=45000,
    department="IT"


)
```
