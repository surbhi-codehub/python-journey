```python
#1.
for i in range(7):
  for j in range(7):
    if i == 0 or i == 6 or j == 0 or j == 6:
      print("*", end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
#2.
for i in range(1, 7):
  for j in range(i):
    print("*", end=" ")
  print()
```

```python
#3.
for i in range(6, 0, -1):
  for j in range(i):
    print("#", end=" ")
  print()
```

```python
#4.
for row in range(1, 7):
  for col in range(1, 7):
    if row % 2 == 0 and col % 2 == 0:
      print((row, col))
```

```python
#5.
for i in range(5):
  for j in range(5):
    if i == 0 or i == 4 or j == 0 or j == 4:
      print("X", end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
#6.
for i in range(1, 11):
  for j in range(1, 11):
    print(i * j, end="\t")
  print()
```

```python
#7.
for i in range(1, 7):
  for j in range(i):
    print(i, end="")
  print()
```

```python
#8.
for i in range(5):
  for j in range(5):
    if (i + j) % 2 == 0:
      print("X", end=" ")
    else:
      print("0", end=" ")
  print()
```

```python
#9.
for row in range(1, 6):
  for col in range(1, 6):
    if row < col:
      print((row, col))
```

```python
#10.
for i in range(1, 6):
  for j in range(i):
    print("*", end=" ")
  print()
```
