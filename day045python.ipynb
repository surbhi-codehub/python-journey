```python
#1 Write a simple while loop.
i = 1

while i <= 5:
  print(i)
  i += 1
```

```python
#2 Use the break statment to stop the loop.
i = 1

while i <= 10:
  if i == 5:
    break
  print(i)
  i += 1
```

```python
#3 Print numbers until the loop stops.
i = 1

while i <= 10:
  if i == 7:
    break
  print(i)
  i += 1
```

```python
#4 change the break condition and observe the output.
i = 1

while i <= 10:
  if i == 4:
    break
  print(i)
  i += 1
```

```python
#5 Try the break statment inside a for loop.
for i in range(1, 11):
  if i == 6:
    break
  print(i)
```

```python
#6. Compare  the output with and without break.
for i in range(1, 6):
  print(i)
```
