```python
#1.
for i in range(7):
  for j in range(7):
    print("*", end=" ")
  print()
```

```python
#2.
for i in range(1, 8):
  for j in range(1, i + 1):
    print(j, end=" ")
  print()
```

```python
#3.
for i in range(7, 0, -1):
  for j in range(i):
    print("#", end=" ")
  print()
```

```python
#4.
for i in range(3):
  for j in range(8):
    print("@", end=" ")
  print()
```

```python
#5.
for i in range(1, 6):
  for j in range(1, 6):
    print(i, j)
```

```python
#6.
for i in range(2, 7):
  print("Table of", i)

  for j in range(1, 11):
     print(i, "x", j, "=", i * j)
```
