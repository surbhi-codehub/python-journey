```python
#1.
for i in range(6):
  for j in range(6):
    print("*", end=" ")
  print()
```

```python
#2.
for i in range(5):
  for j in range(8):
    if i == 0 or i == 4 or j == 0 or j == 7:
      print("*", end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
#3.
for i in range(1, 7):
  for j in range(1, i + 1):
    print(j, end=" ")
  print()
```

```python
#4.
for  row in range(1, 6):
  for column in range(1, 6):
    print((row, column))
```

```python
#5.
for i in range(1, 6):
  for j in range(1, 6):
    if i == j:
      print(i, end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
#6.
for i in range(3, 8):
  print("Table of", i)
  for j in range(1, 11):
      print(i, "x", j, "=", i * j)
  print()
```

```python
#7.
for i in range(6):
  for j in range(6):
    if (i + j) % 2 == 0:
      print("x", end=" ")
    else:
      print("o", end=" ")
  print()
```

```python
#8
for i in range(1, 6):
  for j in range(6):
    print(i, end=" ")
  print()
```

```python
#9.
for i in range(6, 0, -1):
  for j in range(i):
    print("#", end=" ")
  print()
```

```python
#10.
for i in range(5):
  for  j in range(5):
    print("@", end=" ")
  print()
```
