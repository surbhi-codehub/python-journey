```python
#1.
def student_info(**kwargs):
  print(kwargs)


student_info(name="Surbhi", age=20)


```

```python
#2.
def student_info(**kwargs):
  print(kwargs)


student_info(name="Surbhi", age=20)
```

```python
#3.
def student_info(**kwargs):
  print(kwargs)


student_info(name="Surbhi", city="Jameshedpur", country="India")
```

```python
#4.
def student_info(**kwargs):
  for key in kwargs:
    print(key)

student_info(name="Surbhi", age=20, city="Ranchi")
```

```python
#5.
def student_info(**kwargs):
  for value in kwargs.values():
    print(value)

student_info(name="Surbhi", age="30", city="Jameshedpur")
```

```python
#6.
def student_info(**kwargs):
  for key, value in kwargs.items():
    print(key, "=", value)

student_info(name="Student", age=20, city="Ranchi")
```

```python
#7.
def student_info(**kwargs):
    print("Number of arguments:", len(kwargs))

student_info(name="Surbhi", age=20, city="Jamshedpur")
```

```python
#8.
def student_info(**kwargs):
  if "city" in kwargs:
    print("city exsits")
  else:
    print("city does not exist")

student_info(name="Surbhi", age=20, city="Gumla")
```

```python
#9.
def student_info(**kwargs):
  message = f"My name is {kwargs['name']} and I live in {kwargs['city']}"
  print(message)

student_info(name="Surbhi", city="Gumla")
```

```python
#10.
def numbers(*args):
  for value in args:
    print(value)

numbers(10,20,30,40,50)
```
