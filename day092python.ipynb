```python
#1.
def area_rectangle(length, width):
  return length * width

result = area_rectangle(10, 5)
print("Area of rectangle:", result)
```

```python
#2.
def temprature_check(temp):
  if temp >=30:
    return "Hot"

  else:
    return "Cold"

print(temprature_check(35))
```

```python
#3.
def word_length(word):
  return len(word)


result = word_length("Python")
print("Length:", result)
```

```python
#4.
def smallest(a, b, c):
  return min(a, b, c)

result = smallest(16, 78, 89)
print("Smallest number:", result)
```

```python
#5.
def greet(name="student"):
  return "Hello " + name

print(greet())
print(greet("Surbhi"))
```

```python
#6.
def smallest_value(*args):
  return min(args)

result = smallest_value(34, 53, 343, 34, 34)
print("Smallest value:", result)
```

```python
#7.
def show_details(**kwargs):
  for key, value in kwargs.items():
    print(key, ":", value)



show_details(name="Surbhi", age=30, city="Ranchi")
```

```python
#8.
def average_marks(name, mark1, mark2, mark3):
  average = (mark1 + mark2 + mark3) / 3
  return average


result = average_marks("surbhi", 80, 70, 34)
print("Average marks:", result)
```

```python
#9.
def add(a, b):
  return a + b


result = add(10, 39)

print("Returned value:", result)
```

```python
#10.
def product(a, b):
  return a * b


result = product(6, 7)
print("Product:", result)
```
