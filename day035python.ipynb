```python
# Topic: Nested for Loop
```

```python
#1 Wirte an outer for loop.
for i in range(3):
  print(i)
```

```python
#2 Add another for loop inside the first loop.
for i in range(3):
  for j in range(2):
    print(j)
```

```python
#3 Print a simple message from the inner loop.
for i in range(3):
  for j in range(2):
    print("Hello")
```

```python
#4 Observe how many times the inner loop runs.
for i in range(4):
  for j in range(3):
    print("Python")
```

```python
#5 Change the range values and compare the output.
for i in range(2):
  for j in range(4):
    print("Hello")
```

```python
#6 Experiment with different loop counts.
for i in range(3):
  for j in range(3):
    print("Nested Loop")
```

```python
#7 Print numbers from 10 to 1 using a for loolp and range().
for i in range(10, 0, -1):
  print(i)
```
