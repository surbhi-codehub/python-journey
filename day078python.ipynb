```python
#1.
n = 5

# Top half
for i in range(1, n + 1):
  print(" " * (n - i) + "* " * i)


# Bottom half
for i in range(n - 1, 0, -1):
 print(" " * (n - i) + "* " * i)
```

```python
#2.
num = 1

for i in range(1, 6):
  for j in range(1, i + 1):
    print(num, end=" ")
    num += 1
  print()
```

```python
#3.
for i in range(7):
  for j in range(7):
    if i == j:
      print("x", end=" ")
    else:
      print(".", end=" ")
  print()
```

```python
#4.
for row in range(1, 6):
  for column in range(1, 6):
    if row % 2 != 0 and column % 2 != 0:
      print((row, column))
```

```python
#5.
for i in range(5, 10):
  print("Multiplication table of", i)
  for j in range(1, 11):
    print(i, "x", j, "=", i * j)
  print()
```

```python
#6.
n = 5

for i in range(1, n + 1):
  print(" " * (n -i) + (str(i) + " ") * i)
```

```python
#7.
for i in range(6):
  for j in range(6):
    if (i + j) % 2 == 0:
      print("x", end=" ")
    else:
      print("o", end=" ")
  print()
```

```python
#8.
n= 5

for i in range(1, n + 1):
  for j in range(1, i + 1):
    if j == 1 or j == i or i == n:
      print("#", end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
#9.
n = 5

for i in range(1, n + 1):
  for j in range(1, i + 1):
    print(j, end=" ")
  print()
```

```python
#10.
n = 5

for i in range(n):
  for j in range(n):
    if i == 0 or i == n - 1 or j == 0 or j == n - 1:
      print("*", end=" ")
    else:
      print(" ", end=" ")
  print()
```
