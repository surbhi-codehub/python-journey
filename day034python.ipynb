```python
#1 Write a for loop to count backwards using range().
for i in range(10, 0, -1):
  print(i)
```

```python
#2 Print numbers from 10 to 1.
for i in range(10, 0, -1):
  print(i)
```

```python
#3 Change the starting number and observe the output.
for i in range(15, 0, -1):
  print(i)
```

```python
#4 Change the ending number and observe the ouput.
for i in range(10, 4, -1):
   print(i)
```
