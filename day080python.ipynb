```python
#4.
for i in range(8):
  for j in range(8):
    if i == 0 or i == 7 or j == 0 or j == 7:
      print("*", end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
#5.
for i in range(1, 11):
  print("Table of", i)
  for j in range(1, 11):
      print(i, "x", "=", i * j)
  print()
```

```python
#6.
for i in range(1, 9):
  for j in range(1, i + 1):
    print(j, end=" ")
  print()
```

```python
#7.
for row in range(1, 6):
  for column in range(1, 6):
    print("(", row, ",", column, ")")
```
