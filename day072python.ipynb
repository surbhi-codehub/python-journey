```python
#1.
for i in range(5):
  for j in range(5):
    print("*", end=" ")
  print()
```

```python
#2.
for i in range(5):
    for j in range(5):
        if i == 0 or i == 4 or j == 0 or j == 4:
           print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
#3.
for i in range(1, 7):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
#4.
for i in range(1, 6):
  for j in range(5 -i):
    print(" ", end=" ")
  for j in range(i):
    print("*", end=" ")
  print()
```

```python
#5.
for i in range(1, 6):
  for j in range(5 -i):
    print(" ", end=" ")
  for j in range(2 * i -1):
    print("*", end=" ")
  print()
```

```python
#6.
for i in range(1, 4):
  for j in range(1, 5):
    print(i, j, end="   ")
  print()
```

```python
#7.
for i in range(1, 6):
  for j in range(5):
    print(i, end=" ")
  print()
```

```python
#8.
for i in range(5):
  for j in range(5):
    if (i + j) % 2 == 0:
      print("X", end=" ")
    else:
      print("o", end=" ")
  print()
```

```python
#9.
for i in range(1, 6):
  for j in range(1, i + 1):
    print(j, end=" ")
  print()
```

```python
#10.
for i in range(3):
  for j in range(4):
    print("#", end=" ")
  print()
```
