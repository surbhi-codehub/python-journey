```python
# Topic: Common String Method: Join()
```

```python
#1.
words = ["Python", "is", "easy"]
print(words)
```

```python
#2.
words = ["Python", "is", "easy"]

result = "".join(words)
print(result)
```

```python
#3.
words = ["Python", "is", "easy"]

result = " ".join(words)
print(result)
```

```python
#4.
words = ["Python", "is", "easy"]

result = "-".join(words)
print(result)
```

```python
#5.
words = ["Python", "is", "easy"]

space_result = " ".join(words)
hyphen_result = "_".join(words)

print(space_result)
print(hyphen_result)
```

```python
#6.
fruits = ["Apple", "Banana", "Mango"]

result = ", ".join(fruits)
print(result)
```

```python
#7.
sentence = "Python is very easy"

words = sentence.split()
print(words)
```
