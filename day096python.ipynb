```python
#1.
def total_price(price1, price2, price3):
  return price1 + price2 + price3

result = total_price(100, 200, 100)
print("Total Price:", result)
```

```python
#2.
def average(a, b, c, d, e):
  return (a + b + c + d + e) / 5

result = average(10, 20, 30, 40, 50)
print("Average:", result)
```

```python
#3.
def divisible_by_5(number):
  if number % 5 == 0:
    return True
  else:
    return False

print(divisible_by_5(25))
```

```python
#4.
def celsius_to_farhrenheit(celsius):
     return (celsius * 9/5) + 32

result = celsius_to_farhrenheit(25)
print("Fahrenheit:", result)
```

```python
#5.
def longest_word(word1, word2):
  if len(word1) > len(word2):
    return word1
  else:
    return word2

result = longest_word("Python", "Programming")
print("Longest word:", result)
```

```python
#6.
def average(*args):
  return sum(args) / len(args)

result = average(10, 20, 30,40, 50)
print("Average:", result)
```

```python
#7.
def product_details(**kwargs):
  for key, value in kwargs.items():
    print(key, ":", value)

product_details(
    name="Laptop",
    price=50000,
    brand="Avita"
)
```

```python
#8.
def add(a, b):
  return a + b

def multiply(a, b):
  return a * b

def combine(a, b):
  result1 = add(a, b)
  result2 =multiply(a, b)
  return result1, result2

result = combine(5, 3)
print("Combined Result:", result)
```

```python
#9.
def square(number):
  return number * number

result = square(5)
print("Square:", result)
```

```python
#10.
def smaller(a, b):
  if a < b:
    return a
  else:
    return b

result = smaller(10, 7)
print("Smaller Number:", result)
```
