```python
#1. Write a nested for loop.
for i in range(3):
  for j in range(3):
    print("*", end=" ")
    print()
```

```python
#2 Print a square pattern using the * symbol.
for i in range(4):
  for j in range(4):
    print("*", end=" ")
  print()
```

```python
#3 Change the size of the square pattern.
for i in range(5):
  for j in range(5):
    print("*", end=" ")
  print()
```

```python
#4 Print a rectangle pattern.
for i in range(3):
  for j in range(6):
    print("*", end=" ")
  print()
```

```python
#5 Observe how the outer and inner loops work together.
for i in range(3):
  for j in range(4):
      print("*", end=" ")
  print()
```

```python
#6 Experiment with different pattern sizes.
for i in range(2):
  for j in range(5):
     print("*", end=" ")
  print()
```
