```python
#1 Create two comparison expressions.
a = 10
b = 78
print(a > 5)
print(b < 70)
```

```python
#2 Use the or operator to combine both expression.
a = 90
b = 100
result = (a > 80) or (b < 95)
print(result)
```

```python
#3 Print the result of the or operation.
a = 8
b = 12
print((a > 5) or (b > 20))
```

```python
#4 Make one expression True and the other False.
x = 15
y = 5
print((x > 10) or (y > 10))
```

```python
#5 Make both expession False and observe the output.
x = 5
y = 8
print((x > 10) or (y > 10))
```

```python
#6 Try differnt values and compare the results.
a = 25
b = 30
print((a < 20) or (b > 25))
```
