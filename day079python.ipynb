```python
#1.
for i in range(10):
  for j in range(10):
    print("*", end=" ")
  print()
```

```python
#2.
for i in range(6):
  for j in range(10):
    if i == 0 or i == 5 or j == 0 or j == 9:
      print("*", end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
#3.
for i in range(1, 8):
  for j in range(1,i + 1):
    print(j, end=" ")
  print()
```

```python
#4.
for row in range(1, 6):
  for column in range(1, 6):
    if (row + column) % 2 == 0:
      print((row, column))
```

```python
#5.
for i in range(2, 11):
  print("Table of", i)
  for j in range(1, 11):
    print(i, "x", j, "=", i * j)
  print()
```

```python
#6.
for i in range(8):
  for j in range(8):
    if (i +j) % 2 == 0:
      print("*", end=" ")
    else:
      print("#", end=" ")
  print()
```

```python
#7.
for i in range(1, 8):
  for j in range(1, i + 1):
    print(j, end=" ")
  print()
```

```python
#8.
for i in range(5):
  for j in range(5):
    if i == 0 or i == 4 or j == 0 or j == 4:
      print("#", end=" ")
    else:
      print(".", end=" ")
  print()
```

```python
#9.
for i in range(7, 0, -1):
  for j in range(1, i + 1):
    print(j, end=" ")
  print()
```
