```python
#1.
def full_name(first_name, last_name):
  print(first_name, last_name)

full_name("Surbhi", "Kumari")
```

```python
#2.
def introduction(name, age):
  print("My name is", name)
  print("My age", age)

introduction("Surbhi", 20)
```

```python
#3.
def difference(a, b):
  return a - b

result = difference(20, 8)
print(result)
```

```python
#4.
def rectangle_area(length, width):
  return length * width

area = rectangle_area(10, 5)
print("Area =", area)
```

```python
#5
def student_details(name, marks):
  print("Studnet Name:", name)
  print("Marks:", marks)

student_details("Surbhi", 85)
```

```python
#6.
def favorite_colours(color1, colour2, color3):
  print("Favourite colour:")
  print(color1)
  print(colour2)
  print(color3)

favorite_colours("pink", "Blue", "Black")
```

```python
#7.
def greet(name):
  print("Hello", name)

greet("Surbhi")
greet("Rahul")
greet("Priya")
```
