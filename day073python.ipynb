```python
# Topic: More nested Loop Challenges
```

```python
#1.
for i in range(6):
  for j in range(6):
    print("*", end=" ")
  print()
```

```python
#2.
for i in range(1, 7):
  for j in range(1, i + 1):
    print(j, end=" ")
  print()
```

```python
#3.
for i in range(6, 0, -1):
  for j in range(i):
    print("*", end=" ")
  print()
```

```python
#4.
for i in range(5):
  for j in range(8):
    print("$", end=" ")
  print()
```

```python
#5.
for i in range(1, 4):
  for j in range(1, 5):
    print(f"({i}, {j})", end=" ")
  print()
```

```python
#6.
for i in range(1, 6):
  for j in range(5):
    print(i, end=" ")
  print()
```

```python
#7.
for i in range(1, 6):
  for j in range(1, 6):
    print(i * j, end="\t")
  print()
```

```python
#8.
for i in range(5):
    for j in range(6):
        if j % 2 == 0:
           print("A", end=" ")
        else:
           print("B", end=" ")
    print()
```

```python
#9.
for i in range(1, 5):
  for j in range(1, 5):
    print((i, j))
```

```python
#10.
for i in range(4):
  for j in range(4):
    print("#", end=" ")
  print()
```
