```python
# Topic: Nested Loops(Practice).
```

```python
#1.
for i in range(3):
  for j in range(3):
    print("*", end= " ")
  print()
```

```python
#2.
for i in range(4):
  for j in range(4):
    print(j +1, end=" ")
  print()
```

```python
#3.
for i in range(5):
  for j in range(5):
    print("#", end=" ")
  print()
```

```python
#4.
for i in range(1, 6):
  for j in range(1, 6):
    print(i * j, end=" ")
  print()
```

```python
#5.
for row in range(1, 4):
  for column in range(1, 4):
    print(row, column)
```

```python
#6.
for i in range(1, 6):
  for j in range(i):
    print("*", end=" ")
  print()
```

```python
#7.
count = 0

for i in range(4):
  for j in range(3):
    count += 1

print(count)
```

```python
#8.
for i in range(4):
  for j in range(6):
    print("@", end=" ")
  print()

```

```python
#9.
for i in range(1, 6):
  print(i, end= " ")
  for j in range(5):
    print("*", end= " ")
  print()
```

```python
#10.
for i in range(10, 0, -1):
  print(i)
```
