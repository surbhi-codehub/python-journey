```python
#5.
def larger(a, b):
  if a > b:
    return a
  else:
    return b

result = larger(25, 15)
print(result)
```

```python
#7.
def square(number):
  return number * number

print(square(2))
print(square(5))
print(square(10))
```

```python
#9.
def is_even(n):
  if n % 2 == 0:
    return True
  else:
    return False

print(is_even(10))
print(is_even(7))
```
