```python
# Topic: Right Triangle Pattern using Nested for Loops
```

```python
#1. Write a nested for loop.
for i in range(1, 6):
  for j in range(1, 6):
    print(j, end=" ")
  print()
```

```python
#2 Print a right triangle pattern using *
for i in range(1, 6):
  for j in range(i):
    print("*", end=" ")
  print()
```

```python
#3 Increase the number of rows
for i in range(1, 10):
  for j in range(i):
    print("*", end=" ")
  print()
```

```python
#4 Decrease the number of rows
for i in range(1, 4):
  for j in range(i):
    print("*", end=" ")
  print()
```

```python
#5 Use another symbol
for i in range(1, 6):
  for j in range(i):
    print("#", end=" ")
  print()
```

```python
#6 Experiment with different triangle sizes.
for i in range(1, 8):
  for j in range(i):
    print("*", end=" ")
  print()


  for i in range(1, 5):
    for j in range(i):
      print("*", end=" ")
    print()
```

```python
#7 Print a square pattern using nested for loops.
for i in range(5):
  for j in range(5):
    print("*", end=" ")
  print()
```
